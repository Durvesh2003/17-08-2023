```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
df = pd.read_csv('/content/Banking.csv')
df.head(5)
```

```python
df.shape
```

```python
df.info()
```

```python
# Generative descriptive Statistics for thr dataframe
df.describe()     # Shows all the statistics of numerical columns of the numerical data(like,count,max,min,etc+)
```

```python
df['Estimated Income'].min()
```

```python

bins = [0,100000,300000,float('inf')]  # Estimed Income values are grouped down into bins(betwwen 0 to 100000, 100000 to 300000 and 300000 above)
labels = ['low','mid','high']
df['Income Band'] = pd.cut(df['Estimated Income'], bins=bins, labels=labels, right=False)
```

```python
df['Income Band'].value_counts().plot(kind='bar')
```

```python
# Examin the distribution of unique categories in categorical column
categorical_cols = df[["BRId","GenderId","IAId","Amount of Credit Cards","Nationality","Occupation","Fee Structure","Loyalty Classification","Properties Owned","Risk Weighting","Income Band"]].columns
for col in categorical_cols:
    print(f"Value counts for '{col}' :")
    display(df[col].value_counts())
```

Univariate Analysis

```python
for i, predictor in enumerate(df[["BRId","GenderId","IAId","Amount of Credit Cards","Nationality","Occupation","Fee Structure","Loyalty Classification","Properties Owned","Risk Weighting","Income Band"]].columns):
    plt.figure(i)
    sns.countplot(data=df,x=predictor)
```

Bivariate Analysis  (female/Male)


```python
for i, predictor in enumerate(df[["BRId","GenderId","IAId","Amount of Credit Cards","Nationality","Occupation","Fee Structure","Loyalty Classification","Properties Owned","Risk Weighting","Income Band"]].columns):
    plt.figure(i)
    sns.countplot(data=df,x=predictor,hue = 'GenderId')
```

```python
for i, predictor in enumerate(df[["BRId","GenderId","IAId","Amount of Credit Cards","Nationality","Occupation","Fee Structure","Loyalty Classification","Properties Owned","Risk Weighting","Income Band"]].columns):
    plt.figure(i)
    sns.countplot(data=df,x=predictor,hue = 'Nationality')
```

```python
# Histplot for value count for different occupations

for col in categorical_cols:
  if col == "Occupation":
    continue
  plt.figure(figsize = (8,4))
  sns.histplot(df[col])
  plt.title("histogram ofOccupation Count")
  plt.xlabel(col)
  plt.ylabel("Count")
  plt.show()
```
